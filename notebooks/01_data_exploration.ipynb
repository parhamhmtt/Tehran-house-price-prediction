```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
df = pd.read_csv("../data/raw/tehran_house_prices.csv")

print(f"Dataset shape: {df.shape}")
```

```python
display(df.head())
display(df.tail())
```

```python
df.info()
```

```python
print("Columns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python
print("Data types:")
display(df.dtypes.to_frame("dtype"))
```

```python
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": df.isna().mean() * 100
})

display(missing)
```

```python
duplicate_count = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_count}")
print(f"Duplicate percentage: {duplicate_count / len(df) * 100:.2f}%")
```

```python
expected_numeric = [
    "Area",
    "Room",
    "Parking",
    "Warehouse",
    "Elevator",
    "Price",
    "Price(USD)"
]

print("Expected numerical columns:\n")

for col in expected_numeric:
    print(f"{col:15} -> dtype: {df[col].dtype}")
```

```python
print("Non-numeric values in columns expected to be numeric:\n")

for col in expected_numeric:
    converted = pd.to_numeric(df[col], errors="coerce")

    original_missing = df[col].isna().sum()
    converted_missing = converted.isna().sum()

    non_numeric = converted_missing - original_missing

    print(f"{col:15} -> {non_numeric} non-numeric values")
```

```python
numeric_audit = pd.DataFrame()

for col in expected_numeric:
    numeric_audit[col] = pd.to_numeric(df[col], errors="coerce")

display(numeric_audit.describe().T)
```

```python
categorical_cols = df.select_dtypes(include=["object", "str"]).columns

print("Categorical columns:")
print(list(categorical_cols))

for col in categorical_cols:
    print("\n" + "=" * 60)
    print(f"Column: {col}")
    print(f"Unique values: {df[col].nunique(dropna=True)}")
    print("\nTop 20 values:")
    print(df[col].value_counts(dropna=False).head(20))
```

```python
binary_features = ["Parking", "Warehouse", "Elevator"]

for col in binary_features:
    print("\n" + "=" * 50)
    print(col)
    print(df[col].value_counts(dropna=False).sort_index())
```

```python
for col in ["Area", "Room", "Price", "Price(USD)"]:
    values = pd.to_numeric(df[col], errors="coerce")

    print(f"\n===== {col} =====")
    print(f"Values <= 0: {(values <= 0).sum()}")
    print(f"Missing values: {values.isna().sum()}")
```

```python
for col in ["Area", "Room", "Price", "Price(USD)"]:
    values = pd.to_numeric(df[col], errors="coerce")

    print("\n" + "=" * 60)
    print(f"{col}")

    print("\nSmallest 10:")
    print(values.nsmallest(10).tolist())

    print("\nLargest 10:")
    print(values.nlargest(10).tolist())
```

```python
print("Number of unique addresses:", df["Address"].nunique(dropna=True))

print("\nTop 30 addresses:")
display(df["Address"].value_counts(dropna=False).head(30))

print("\nFirst 30 addresses:")
display(df["Address"].head(30).tolist())
```

```python
missing_sorted = missing.sort_values(
    by="missing_percentage",
    ascending=False
)

display(missing_sorted)
```

```python
print("Basic consistency checks")
print("=" * 50)

print(f"Rows: {len(df)}")
print(f"Columns: {len(df.columns)}")

print(f"\nArea unique values: {df['Area'].nunique(dropna=True)}")
print(f"Address unique values: {df['Address'].nunique(dropna=True)}")

print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

```python
print("=" * 70)
print("DAY 1 DATASET AUDIT SUMMARY")
print("=" * 70)

print(f"Dataset size: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"Duplicate rows: {df.duplicated().sum()}")

print("\nMissing values:")
for col in df.columns:
    count = df[col].isna().sum()
    if count > 0:
        print(f"  {col}: {count} ({count / len(df) * 100:.2f}%)")

print("\nExpected numerical columns:")
for col in expected_numeric:
    converted = pd.to_numeric(df[col], errors="coerce")
    non_numeric = converted.isna().sum() - df[col].isna().sum()
    print(f"  {col}: {non_numeric} non-numeric values")

print("\nCategorical columns:")
for col in categorical_cols:
    print(f"  {col}: {df[col].nunique(dropna=True)} unique values")

print("\nInvalid values:")
for col in ["Area", "Room", "Price", "Price(USD)"]:
    values = pd.to_numeric(df[col], errors="coerce")
    print(f"  {col} <= 0: {(values <= 0).sum()}")
```
